```python
# to calculate percentage and display highest and lowest marks .
student_name = input("Enter student name:")
roll_no = input("Enter roll no:")
# to get subject and amrks from user
Subject1 = input("Enter name of Subject1:")
Marks1 = int(input("Enter marks for {Subject1}:"))
Subject2 = input("Enter name of Subject2:")
Marks2 = int(input("Enter marks for {Subject2}:"))
Subject3 = input("Enter name of Subject3:")
Marks3 = int(input("Enter marks for {Subject3}:"))
# to calculate percentage
total_marks = Marks1 + Marks2 + Marks3
percentage = (total_marks/300)*100
# to calculate lowest and highest marks
Marks = {Subject1: Marks1,Subject2: Marks2,Subject3: Marks3}
highest = max(Marks)
lowest = min(Marks)
# display of student report card 
print("Student marks:", student_name)
print("Roll number:", roll_no)
print("Total marks:", total_marks)
print("Percentage:", percentage,"%")
print("Highest Marks:",highest,"=",Marks[highest])
print("Lowest Marks:",lowest,"=",Marks[lowest])
```

```python
# TO WRITE PYTHON CODE TO CHECK WHETHER GIVEN NUMBER IS POSITIVE , NEGATIVE OR ZERO
Number = int(input("Enter your number :"))
if ( Number>0 ):
    print("Number is positive")
elif ( Number<0 ):
    print("Number is negative")
else:
    print("Number is zero")
```

```python
# TO WRITE PPYTHON CODE TO CHECKWHETHER NUMBER S ODD OR EVEN 
Number = int(input("Enter your number:"))
if ( Number%2==0 ):
    print(" Number is even ")
else:
    print(" Number is odd ")
```

```python
# TO CHECK GIVEN NUMBER IS PRIME OR NOT 
number = int(input("Enter your number:"))
if ( number%1==0 and number%number==0 ):
    print("Number is Prime")
else:
    print("Number is not prime")
```

```python
# to write python program to print the sum of digits of number 
n =input("Enter your number:")
sum = 0
for i in n:
    sum = sum + int(i)
print("sum of digits:",sum)
```

```python
# to reverse a number and print 
num = input("Enter your number:")
reversed_num = int(num[ : :-1])
print(reversed_num)
```

```python
num = input("enter your number:")
reversed_num = (num[ : :-1])
print(reversed_num)
if ( num == reversed_num ):
    print("Palindrome number")
else:
    print("Not Palindrome number")
```
