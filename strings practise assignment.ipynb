```python
# WRITE A PROGRAM TO COUNT THE NO OF UPPER CASE AND LOWER CSE LETTER IN STRING
s = input("Enter your string:",)
upper = 0
lower = 0
for i in s:
    if s.isupper():
        upper += 1
    elif i.islower():
        lower += 1
print("Upper_case:",upper)
print("Lower_case:",lower)
```

```python
# WRITE A PRIOGRAM TO CHECK WHETHER A GIVEN STRING IS PALINDROME OR NOT
s = input("Enter your string:")
print("String:",s)
if s == s[: :-1]:
    print("String is palindeome")
else:
    print("Not palindrome")
```
