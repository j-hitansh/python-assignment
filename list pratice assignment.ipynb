```python
# ACCEPT 20 VALUES AND PERFORM OPERATIONS 
Numbers = []
for i in range(20):
    value = int(input("Enter your value:"))
    Numbers.append(value)
print("List:",Numbers)
# to count even and odd numbers
even = 0
odd = 0
for j in Numbers:
    if (j%2==0 ):
        even+=1
    else:
        odd+=1
print("Even values:",even)
print("Odd values:",odd)
```

```python
# sort list in ascending order 
Numbers = []
for i in range(10):
    values = int(input("Enter your values:"))
    Numbers.append(values)
print("List:",Numbers)
ascending = sorted(Numbers)
print("Ascending order:",ascending)
```

```python
# sort list in descending order 
Numbers = []
for i in range(10):
    values = int(input("Enter your values:"))
    Numbers.append(values)
print("List:",Numbers)
Numbers.sort(reverse=True)
print("Descending order:",Numbers)
```
