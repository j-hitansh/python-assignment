```python
# WRITE A PROGRAM TO CONCENTATE FOOLOWINF DICTIONARY TO CREATE A NEW ONE 
dic1 = {1:10,2:20}
print("Dic:",dic1)
dic2 = {3:30,4:40}
print("Dic:",dic2)
dic3 = {5:50,6:60}
print("Dic:",dic3)
dic = {}
dic.update(dic1)
dic.update(dic2)
dic.update(dic3)
print("Concentenated Dic:",dic)
```

```python
# WRITE A PROGRAM ITERATE OVER A DICTIONARY USING FOR LOOP AND PRINT THE KEYS ALONE,VALUES ALONE AND BOTH KEYS AND VALUES 
dictionary = {"NAME":"HITANSH JAIN","AGE":19,"BIRTH DATE":19/2/2008}
print("Dictionary:",dictionary)
for key in dictionary.keys() :
    print("Keys:",key)
for value in dictionary.values():
    print("value:",value)
for item in dictionary.items():
    print("Items:",item)
```

```python
dictionary = {"NAME":"HITANSH JAIN","AGE":18,"PRN":1272260661,"BRANCH":"CSE","YEAR":"FIRST YEAR"}
print("Dictionary:",dictionary)
# update the information 
dictionary.update({"BRANCH":"CSE (CLOUD COMPUTING)"})
print("Dictionary afer updating:",dictionary)
# adding the element 
dictionary["DIVISION"]=14
print("Dictionaary after adding:",dictionary)
# remove the element 
dictionary.pop("YEAR")
print("Dictionary after delete:",dictionary)
```

```python
# TO FIND MEAN OF VALUES OF DICTIONARY 
my_dict = {"MARKS1":87,"MARKS2":89,"MARKS3":91,"MARKS4":95,"MARKS5":100}
print("My_Dict:",my_dict)
Mean = sum(my_dict.values())/5
print("Mean of all values:",Mean)
```

```python
#  SORT THE ELEMENTS OF DICTIONARY USING SORTED FUNCTIONS 
dictionary = {"NAME":"HITANSH JAIN","PANEL":"CSE","ROLL NO":1272260661,"MARKS":[65,87,67,94]}
print("Dictionary:",dictionary)
# sorting the keys 
Sorting_keys = sorted(dictionary.keys())
print("Sorted keys:",Sorting_keys)
Sorting_values = sorted(dictionary.values(),key=str)
print("Sorted values:",Sorting_values)
```

```python
# CONVERT DICTIONARY TO LIST. ONE LIST OF KEYS AND ONE LIST OF VALUES.
my_dict = {"NAME":"HITANSH JAIN","PANEL":"CSE","ROLL NO":1272260661}
print("MY_Dict:",my_dict)
list1 = list(my_dict.keys())
list2 = list(my_dict.values())
print("List1:",list1)
print("List2:",list2)
```

```python
# CONVERT LIST INTO DICTIONARY.
list1 = ("NAME","PANEL","ROLL NO")
print("LIST1:",list1)
list2 = ("HITANSH JAIN","CSE",1272260661)
print("LIST2:",list2)
my_dict = dict(zip(list1,list2))
print("My_dict:",my_dict)
```
