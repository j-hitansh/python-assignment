```python
# CREATE A 3 CROSS 3 ARRAY OF ALL TRUE
import numpy as np
array = np.ones((3,3),dtype = bool)
print("New_array:",array)
```

```python
# CREATE A 1D ARRAY OF 10 EVENLY SPACED VALUES BETWEEN 5 AND 50 
import numpy as np
array = np.linspace(5,50,10)
print("new_array:",array)
```

```python
# CONVERT A PYTHON LIST INTO ARRAY.
import numpy as np
list = [10,20,30,40,50]
print("List:",list)
array = np.array(list)
print("New_array:",array)
```

```python
# REVERSE A 1D NUMPY ARRAY
import  numpy as np
array = np.array([10,20,30,40,50])
print("Array:",array)
reversed_array = array[ : :-1]
print("New_array:",reversed_array)
```

```python
# CREATE 3 CROSS 3 IDENTITY MATRIX
import numpy as np
array = np.eye(3)
print("Matrix:",array)
```

```python
# CREATE 4 CROSS 4 ARRAY AND COMPUTE FIRST ROW AND last  COLUMN
import numpy as np 
array = np.arange(1,17).reshape(4,4)
print("Array:",array)
first_row = array[0, :]
print("First_row:",first_row)
last_column = array[:, -1]
print("Last_column:",last_column)
```

```python
#  CREATE 4 CROSS 4 ARRAY AND COMPUTE FIRST TWO ROW AND LAST TWO COLUMN 
import numpy as np 
arr = np.arange(1,17).reshape(4,4)
print("Array:",arr)
first_row = arr[:2, :2]
print("First_row:", first_row)
first_column = arr[:2, :2]
print("First_column:",first_column)
```

```python
# PERFORM ARITHEMATIC OPERATIONS 
import numpy as np
a =  np.array([10,20,30])
b = np.array([18,24,32])
print("Array_A:",a)
print("Array_B:",b)
print("Addition:",a+b)
print("Subtraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b)
```

```python
#  TO CALCULATE THE DOT PRODUCT OF TWO ARRAY
import numpy as np
a = np.array([10,20,30])
b = np.array([40,50,60])
print("Array_A:",a)
print("Array_B:",b)
c = np.dot(a,b)
print("Dot product:",c)
```

```python
# COMPUTE MEAN ,MEDIAN AND STANDARD DEVIATION OF NUMPY ARRAY
import numpy as np
a = np.array([10,20,30,40,50])
print("Array_A:",a)
mean = np.mean(a)
median = np.median(a)
std = np.std(a)
print("Mean:",mean)
print("Median:",median)
print("Std:",std)
```
