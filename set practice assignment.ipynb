```python
# program to remove alement in set 
set = {"badminton","table tennis","football","foosball"}
print("Set:",set)
item_to_remove = "table tennis"
set.remove(item_to_remove)
print("New_set",set)
```

```python
# program to crate an union of sets
set1 = {"football","cricket"}
print("Set1:",set1)
set2 = {"carrom","badminton","kabbadi"}
print("Set2:",set2)
new_set = set1.union(set2)
print("New_set:",new_set)
```

```python
# program to create an intersection of sets 
set1 = {"football","cricket","basketball","volleyball"}
print("Set1:",set1)
set2 = {"carrom","football","cricket","basketball"}
print("Set2:",set2)
new_set = set1.intersection(set2)
print("New_set:",new_set)
```
