```python
# write python program to check the that a string ccontains only a set of characters in this case a-z,A-Z and 0-9
import re 
def validate_string(user_input):
    pattern = r'^[a-zA-Z0-9]+$'
    if re.match(pattern,user_input):
        print( "Valid! The string containonly a-z, A-z, 0-9.")
    else:
        print("Invalid! The string contain not allowed characters or symbols.")
String = input("Enter the string to check:")
result = validate_string(String)
print(result)
```
