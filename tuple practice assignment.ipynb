```python
#  WRITE PROGRAMT TO PRINT 4 ELEMENT FROM FIRST AND 4 ELEMENTT FROM LAST 
t = (10,20,30,40,50)
print("Tuple:",t)
print("4th eelemt from first:",t[3])
print("4th element from last:",t[-4])
```

```python
#  WRITE PROGRAM TO CHECK WHETHER ELEMENT PRESENT IN THE TUPLE
t = ("apple","banana","grapes","mango")
print("Tuple:",t)
if ( "banana" in t ):
    print("Yes")
else:
    print("Not")
```

```python
# TO CONVERT A LIST INTO TUPLE
Numbers = [10,20,30,40,50,60]
print("List:",Numbers)
t = tuple(Numbers)
print("Tuple:",t)
```

```python
# WRITE A PROGRAM TO FIND INDEX OF AN ITEM IN A TUPLE
t = (10,20,30,40,50,60)
i = int(input("Enter your i:"))
if i in t:
    print("Index of",i,"is",t.index(i))
else:
    print("Item not found")
```
